```python
import yfinance as yf
import pandas as pd
from datetime import datetime, timedelta

tickers = ["TSLA", "AMZN", "AAPL"]

# define 730-day chunks (max allowed)
periods = [
    ("2022-01-01", "2022-12-31"),
    ("2023-01-01", "2023-12-31"),
    ("2024-01-01", "2024-12-31"),
    ("2025-01-01", "2025-12-31"),
]

dfs = {ticker: [] for ticker in tickers}

for ticker in tickers:
    for start, end in periods:
        try:
            df = yf.download(ticker, start=start, end=end, interval="1h")
            if not df.empty:
                dfs[ticker].append(df)
                print(f"✓ {ticker} {start} to {end}: {len(df)} rows")
        except Exception as e:
            print(f"✗ {ticker} {start}-{end}: {e}")

# combine
df_tsla = pd.concat(dfs["TSLA"]) if dfs["TSLA"] else None
df_amzn = pd.concat(dfs["AMZN"]) if dfs["AMZN"] else None
df_aapl = pd.concat(dfs["AAPL"]) if dfs["AAPL"] else None

print(df_tsla.shape, df_amzn.shape, df_aapl.shape)
```

```python
from datasets import load_dataset, Dataset
from collections import deque

ds_stream = load_dataset(
    "Brianferrell787/financial-news-multisource",
    split="train",
    streaming=True,
    download_config=download_config
)

# First, collect all rows in a buffer (streaming from back requires this)
# We use a deque with maxlen to only keep the last N rows if memory is a concern
BUFFER_SIZE = 500000  # Adjust based on your memory — keeps last 500k rows
print("Buffering tail of dataset...")

buffer = deque(maxlen=BUFFER_SIZE)
total_rows = 0

for row in ds_stream:
    buffer.append(row)
    total_rows += 1
    if total_rows % 100000 == 0:
        print(f"Buffered {total_rows:,} rows so far...")

print(f"Done. Total rows: {total_rows:,} | Buffer holds last {len(buffer):,} rows")

# Now scan BACKWARDS through the buffer
target_years = {2022, 2023, 2024, 2025}
results = []
no_match_streak = 0
NO_MATCH_LIMIT = 300000  # Stop if 300k consecutive rows have no match

print("Scanning backwards...")

for row in reversed(buffer):
    if any(str(y) in str(row["date"]) for y in target_years):
        results.append(row)
        no_match_streak = 0  # Reset streak on match
    else:
        no_match_streak += 1

    if no_match_streak >= NO_MATCH_LIMIT:
        print(f"No matches in last {NO_MATCH_LIMIT:,} rows — stopping early.")
        break

# Reverse results so chronological order is preserved
results.reverse()

ds_filtered = Dataset.from_list(results)
print(f"Final dataset: {len(ds_filtered):,} rows")
```

```python
from datasets import load_dataset, Dataset

target_years = {2022, 2023, 2024, 2025}
results = []
no_match_streak = 0
NO_MATCH_LIMIT = 300000
CHUNK_SIZE = 100000  # How many rows to load per chunk

offset = 0  # How far from the end we are

while True:
    start = -(offset + CHUNK_SIZE)
    end   = -offset if offset > 0 else None  # None = up to the very last row

    split = f"train[{start}:]" if end is None else f"train[{start}:{end}]"
    print(f"Loading slice: {split}")

    chunk = load_dataset(
        "Brianferrell787/financial-news-multisource",
        split=split,
        download_config=download_config
    )

    chunk_results = []
    chunk_no_match = 0

    # Scan this chunk backwards
    for row in reversed(chunk.to_list()):
        if any(str(y) in str(row["date"]) for y in target_years):
            chunk_results.append(row)
            no_match_streak = 0
        else:
            no_match_streak += 1
            chunk_no_match += 1

        if no_match_streak >= NO_MATCH_LIMIT:
            print(f"No matches in {NO_MATCH_LIMIT:,} consecutive rows — stopping.")
            break

    results.extend(reversed(chunk_results))  # keep chronological order
    offset += CHUNK_SIZE

    if no_match_streak >= NO_MATCH_LIMIT:
        break

ds_filtered = Dataset.from_list(results)
print(f"Final dataset: {len(ds_filtered):,} rows")
```
